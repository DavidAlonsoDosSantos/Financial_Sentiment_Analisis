# 01 — Exploratory Data Analysis
This notebook explores the Financial PhraseBank dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt
from datasets import load_dataset
```

```python
dataset = load_dataset('takala/financial_phrasebank', 'sentences_allagree')
df = dataset['train'].to_pandas()
df.head()
```

```python
df = df.rename(columns={'sentence': 'text'})
label_map = {0: 'negative', 1: 'neutral', 2: 'positive'}
df['sentiment'] = df['label'].map(label_map)
df.head()
```

```python
df.info()
df['sentiment'].value_counts()
```

```python
df['sentiment'].value_counts().plot(kind='bar')
plt.title('Sentiment Class Distribution')
plt.xlabel('Sentiment')
plt.ylabel('Number of sentences')
plt.tight_layout()
plt.show()
```

```python
df['text_length'] = df['text'].str.split().apply(len)
df['text_length'].plot(kind='hist', bins=30)
plt.title('Text Length Distribution')
plt.xlabel('Number of words')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```
