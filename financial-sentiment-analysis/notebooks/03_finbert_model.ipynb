# 03 — FinBERT Model
FinBERT is a BERT-based model adapted for financial language.

```python
from transformers import pipeline
```

```python
classifier = pipeline('sentiment-analysis', model='ProsusAI/finbert', tokenizer='ProsusAI/finbert')
```

```python
examples = [
    'The company reported stronger-than-expected quarterly earnings.',
    'The firm warned investors about declining revenue and weaker demand.',
    'The board announced that the annual meeting will take place next month.'
]
for text in examples:
    print(text)
    print(classifier(text)[0])
    print()
```

## Notes for presentation
FinBERT is used because financial text has domain-specific vocabulary and sentiment patterns.
